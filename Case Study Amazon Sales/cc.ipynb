```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
df = pd.read_csv("amazon_Sales.csv")
```

```python
df.head()
```

```python
df.info() 
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df["rating_count"]=df["rating_count"].fillna(0)
df["discounted_price"] =( df["discounted_price"].str.replace("₹","",regex=False).str.replace(",","",regex=False).astype(float))
df["actual_price"] = (df["actual_price"].str.replace("₹","",regex=False).str.replace(",","",regex=False).astype(float))
df["discount_percentage"] = (df["discount_percentage"].str.replace("%","",regex=False).astype(float))
df["rating_count"] = (df["rating_count"].str.replace(",","",regex=False).astype(float))
df["rating"]=pd.to_numeric(df["rating"],errors ="coerce")
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df["rating"]=df["rating"].fillna(0)
df["rating_count"] = df["rating_count"].fillna(df["rating_count"].mean())
```

```python
df.isnull().sum()
```

```python
Duplicate_Detected = df.duplicated().sum()
print(f"Duplicate Detected : {Duplicate_Detected}") 
```

```python
num_col = ["discounted_price","actual_price","discount_percentage","rating","rating_count"]
cat_col=df.select_dtypes(include=["object","str"]).columns
```

```python
df[num_col].describe
```

```python
for col in num_col:
    plt.figure(figsize=(6,4))
    plt.hist(df[col],bins=20)
    plt.title("distribution of"+" "+col)
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()
```

```python
for col in num_col:
    plt.Figure(figsize=(6,4))
    sns.boxenplot(x=df[col])
    plt.show()
```

```python
# Scatter Plot

sns.regplot(data=df, x="actual_price", y="discounted_price")
plt.xlabel("Actual_Price")
plt.ylabel("Discounted_Price")
plt.show()
```
